```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
def deterministic_gaurdrail(text:str)->bool:
    banned_keywords=["hack","exploit","malware","phishing","ddos","ransomware","bomb"]
    return any(kw in text.lower() for kw in banned_keywords)


text_inputs=[
    "How to hack into a computer?",
    "What is the best way to learn programming?",
    "Explain the malware spreads"
]

for inp in text_inputs:
    if deterministic_gaurdrail(inp):
        print(f"Input: '{inp}' is flagged by the guardrail.")
    else:
        print(f"Input: '{inp}' is safe.")
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

def model_based_gaurdrail(text:str)->bool:
    llm=ChatOpenAI(model="gpt-4o-mini",temperature=0)
    system_prompt = """
You are a safety classifier.
Answer only 'Yes' or 'No'.

Return 'Yes' if the user's message requests harmful, illegal, malicious, or dangerous instructions.
Otherwise return 'No'.
"""
    response=llm.invoke([SystemMessage(content=system_prompt),HumanMessage(content=text)])
    return response.content.strip().lower()=="yes"



ans=model_based_gaurdrail("How to learn programming?")
print(f"Model-based guardrail response: {'Flagged' if ans else 'Safe'}")
```
