```python
from langchain_community.document_loaders import TextLoader

text_loader=TextLoader("vecless.txt",encoding="utf-8")

loaded_documents=text_loader.load()
```

```python
import re
import json

text = loaded_documents[0].page_content

# Split into lines, detect heading level by number of '#'
heading_pattern = re.compile(r'^(#{1,6})\s+(.*)$')

def parse_tree(text):
    lines = text.split('\n')
    root = {"title": "root", "level": 0, "content": "", "children": []}
    stack = [root]

    buffer = []

    def flush_buffer_to(node):
        node["content"] += "\n".join(buffer).strip()
        buffer.clear()

    for line in lines:
        match = heading_pattern.match(line)
        if match:
            # flush accumulated body text to the current deepest node
            flush_buffer_to(stack[-1])

            level = len(match.group(1))
            title = match.group(2).strip()
            node = {"title": title, "level": level, "content": "", "children": []}

            # pop stack until we find the correct parent (level - 1 or less)
            while stack[-1]["level"] >= level:
                stack.pop()

            stack[-1]["children"].append(node)
            stack.append(node)
        else:
            buffer.append(line)

    flush_buffer_to(stack[-1])
    return root

tree = parse_tree(text)
print(json.dumps(tree, indent=2))
```
